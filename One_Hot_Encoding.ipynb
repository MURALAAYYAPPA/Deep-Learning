```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
df = pd.DataFrame({'color':['blue','red','green','orange','yellow']})
print('original dataframe:\n',df)
print("_"*20)
encoder = OneHotEncoder()
encoded_data = encoder.fit_transform(df[['color']])
encoded_array = encoded_data.toarray()
feature_names = encoder.get_feature_names_out(['color'])
one_hot_df = pd.DataFrame(encoded_array,columns=feature_names)
print("one hot encoded data frame:")
print(one_hot_df)
```

```python
import pandas as pd
df = pd.DataFrame({'course':['AIML','CSE','EEE','ME','ECE','IOT','IT']})
print('original dataframe:',df)
print("_"*20)
one_hot_df_pd = pd.get_dummies(df,columns=['course'],dtype=int)
print("one hot encoded data frame:")
print(one_hot_df_pd)
```

```python
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
sentence = "deep learning is the sub part of machine learning and machine learning is the sub part of AI"
words = sentence.lower().replace('.', '').split()
print(f'tokenized words : {words}')
print("-"*30)
df = pd.DataFrame(words,columns=['word'])
print('original dataframe:\n',df)
print("-"*30)
encoder = OneHotEncoder(sparse_output=False)
encoded_words = encoder.fit_transform(df[['word']])
feature_names = encoder.get_feature_names_out(['word'])
one_hot_df = pd.DataFrame(encoded_words,columns=feature_names)
print("one hot encoded data frame:")
print(one_hot_df)
```
